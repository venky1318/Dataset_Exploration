```python
# Task 17: Dataset Exploration

## Objective
Explore a dataset using Pandas to understand its dimensions,
columns, data types, summary statistics, and basic patterns
before machine-learning modelling.

## Dataset
Iris Dataset
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Convert the dataset into a Pandas DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add the target species as a readable column
df["species"] = [
    iris.target_names[i] for i in iris.target
]

print("Dataset loaded successfully!")
print("Dataset name: Iris Dataset")
```

```python
print("FIRST FIVE RECORDS")
display(df.head())
```

```python
print("LAST FIVE RECORDS")
display(df.tail())
```

```python
print("DATASET DIMENSIONS")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Shape:", df.shape)

print("\nCOLUMN NAMES")
print(df.columns.tolist())
```

```python
print("DATASET INFORMATION")
df.info()
```

```python
print("SUMMARY STATISTICS")
display(df.describe())
```

```python
numerical_columns = df.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = df.select_dtypes(
    exclude="number"
).columns.tolist()

print("NUMERICAL COLUMNS")
print(numerical_columns)

print("\nCATEGORICAL COLUMNS")
print(categorical_columns)
```

```python
print("MISSING VALUES")
display(df.isnull().sum())

print("\nTOTAL MISSING CELLS")
print(df.isnull().sum().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python
print("SPECIES COUNTS")
display(df["species"].value_counts())

print("\nAVERAGE PETAL LENGTH BY SPECIES")
display(
    df.groupby("species")["petal length (cm)"]
      .mean()
      .round(2)
)
```

```python
print("=" * 60)
print("DATASET EXPLORATION: KEY OBSERVATIONS")
print("=" * 60)

print(
    f"1. The dataset contains {df.shape[0]} rows "
    f"and {df.shape[1]} columns."
)

print(
    f"2. There are {len(numerical_columns)} numerical "
    f"columns and {len(categorical_columns)} categorical column."
)

print(
    f"3. The dataset contains "
    f"{df['species'].nunique()} distinct species."
)

print(
    f"4. The total number of missing cells is "
    f"{df.isnull().sum().sum()}."
)

print(
    f"5. The dataset contains "
    f"{df.duplicated().sum()} duplicate rows."
)

species_counts = df["species"].value_counts()

print("\nSpecies distribution:")
print(species_counts.to_string())

species_means = (
    df.groupby("species")["petal length (cm)"]
      .mean()
      .round(2)
)

print("\nAverage petal length by species:")
print(species_means.to_string())

highest_variation = (
    df[numerical_columns].std().idxmax()
)

print(
    f"\n6. '{highest_variation}' has the highest "
    "standard deviation among the numerical columns."
)

print("\nExploration completed successfully.")
```

```python
df.to_csv("iris_dataset.csv", index=False)

print("Dataset exported as iris_dataset.csv")
```
